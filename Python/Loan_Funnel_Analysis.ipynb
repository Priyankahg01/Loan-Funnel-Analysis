```python
call_vs_application = call_df.merge(loan_df[["Customer_ID", "Application_Date"]], left_on = "Customer_ID", right_on = "Customer_ID", how = "left")
call_vs_application["Call_Before_Application"] = (call_vs_application["Call_Date"] < call_vs_application["Application_Date"])
call_vs_application["Call_Before_Application"].value_counts()
```

```python
call_vs_application.loc[call_vs_application["Call_Before_Application"] == True, "Customer_ID"].nunique()
```

```python
pre_call_customers = call_vs_application.loc[call_vs_application["Call_Before_Application"] == True, "Customer_ID"].unique()

loan_df["Had_Pre-Application_Call"] = loan_df["Customer_ID"].isin(pre_call_customers)

pd.crosstab(loan_df["Had_Pre-Application_Call"], loan_df["Loan_Processed"],normalize = "index") * 100
```

```python
pre_call_counts = (call_vs_application[call_vs_application["Call_Before_Application"] == True].groupby("Customer_ID").size())
pre_call_counts.describe()
```

```python
loan_with_calls = loan_df.merge(pre_call_counts.rename("Pre_Application_Call_Count"), left_on="Customer_ID", right_index=True, how = "left")

loan_with_calls["Pre_Application_Call_Count"] = (loan_with_calls["Pre_Application_Call_Count"].fillna(0))

loan_with_calls.groupby("Pre_Application_Call_Count")["Loan_Processed"].mean() * 100
```

```python
completed_calls = (call_df[call_df["Call_Status"] == "Completed"].groupby("Customer_ID").size())

loan_with_calls = loan_df.merge(completed_calls.rename("Completed_Call_Count"), left_on = "Customer_ID", right_index = True, how = "left")

loan_with_calls["Completed_Call_Count"] = (loan_with_calls["Completed_Call_Count"].fillna(0))

loan_with_calls.groupby("Completed_Call_Count")["Loan_Processed"].mean() * 100
```

```python
loan_with_calls["Completed_Call_Count"].value_counts().sort_index()
```

```python
loan_with_calls["Call_Frequency_Group"] = pd.cut(loan_with_calls["Completed_Call_Count"], bins=[-1, 0, 4, 10], labels=["0 Calls", "1-4 Calls", "5-10 Calls"])

loan_with_calls.groupby("Call_Frequency_Group", observed=True)["Loan_Processed"].mean() * 100
```

```python
call_group_summary = loan_with_calls.groupby("Call_Frequency_Group",observed=True).agg(Customers=("Customer_ID", "count"), Processing_Rate=("Loan_Processed", "mean"))

call_group_summary["Processing_Rate"] = (call_group_summary["Processing_Rate"] * 100)

call_group_summary
```

```python
import matplotlib.pyplot as plt

funnel_stages = [
    "Applications",
    "Approved",
    "Processed",
    "Repayment_Date_Recorded"
]

funnel_values = [
    funnel["Applications"],
    funnel["Approved"],
    funnel["Processed"],
    funnel["Repayment_Date_Recorded"]
]

plt.figure(figsize=(10, 6))
plt.plot(funnel_stages, funnel_values, marker="o", linewidth=2)

for i, value in enumerate(funnel_values):
    plt.text(i, value + 20, str(value), ha="center")

plt.title("Loan_Application_Funnel")
plt.xlabel("Funnel_Stage")
plt.ylabel("Number_of_Customers")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
product_processing = (
    loan_df.groupby("Product_Type")["Loan_Processed"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
product_processing.plot(kind="bar")

plt.title("Loan Processing Rate by Product Type")
plt.xlabel("Product_Type")
plt.ylabel("Processing Rate (%)")
plt.xticks(rotation=0)

for i, value in enumerate(product_processing):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.ylim(0, 75)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
credit_processing = (
    loan_df.groupby("Customer_Credit_Score")["Loan_Processed"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
credit_processing.plot(kind="bar")

plt.title("Loan Processing Rate by Credit Score")
plt.xlabel("Customer Credit Score")
plt.ylabel("Processing Rate (%)")
plt.xticks(rotation=0)

for i, value in enumerate(credit_processing):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.ylim(0, 75)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
income_processing = (
    loan_df.groupby("Customer_Income_Level")["Loan_Processed"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
income_processing.plot(kind="bar")

plt.title("Loan Processing Rate by Income Level")
plt.xlabel("Customer Income Level")
plt.ylabel("Processing Rate (%)")
plt.xticks(rotation=0)

for i, value in enumerate(income_processing):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.ylim(0, 75)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
rep_chart = rep_summary.sort_values(
    "Processing_Rate",
    ascending=True
)

plt.figure(figsize=(10, 7))
plt.barh(
    rep_chart.index,
    rep_chart["Processing_Rate"]
)

plt.title("Loan Processing Rate by Sales Representative")
plt.xlabel("Processing Rate (%)")
plt.ylabel("Sales Representative")

for i, value in enumerate(rep_chart["Processing_Rate"]):
    plt.text(value + 0.5, i, f"{value:.1f}%", va="center")

plt.xlim(0, 85)
plt.grid(axis="x", alpha=0.3)

plt.show()
```

```python
call_status_counts = call_df["Call_Status"].fillna("Unknown").value_counts()

plt.figure(figsize=(8, 5))
call_status_counts.plot(kind="bar")

plt.title("Call Outcome Distribution")
plt.xlabel("Call Status")
plt.ylabel("Number of Calls")
plt.xticks(rotation=0)

for i, value in enumerate(call_status_counts):
    plt.text(i, value + 50, str(value), ha="center")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
call_engagement = (
    loan_with_calls.groupby(
        "Call_Frequency_Group",
        observed=True
    )["Loan_Processed"]
    .mean()
    .mul(100)
)

plt.figure(figsize=(8, 5))
call_engagement.plot(kind="bar")

plt.title("Loan Processing Rate by Call Engagement")
plt.xlabel("Completed Call Frequency")
plt.ylabel("Processing Rate (%)")
plt.xticks(rotation=0)

for i, value in enumerate(call_engagement):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.ylim(0, 75)
plt.grid(axis="y", alpha=0.3)

plt.show()
```
